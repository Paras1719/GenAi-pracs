```python
import re
```

```python
spam_keywords = [
    "free", "win", "winner", "won", "lottery", "prize", "cash",
    "reward", "gift", "offer", "discount", "deal", "bonus",
    "urgent", "limited", "exclusive", "guaranteed", "claim",
    "click", "subscribe", "buy", "purchase", "cheap", "sale",
    "money", "earn", "income", "credit", "loan", "investment",
    "bitcoin", "crypto", "million", "selected", "congratulations",
    "risk-free", "refund", "promo", "voucher", "coupon",
    "password", "verify", "account", "bank", "update",
    "act now", "hurry", "instant", "lucky", "gift card"
]

email = input("Enter email text:\n").lower()

email = re.sub(r'[^a-zA-Z\s]', '', email)

count = 0
found = []

for word in spam_keywords:
    if word in email:
        count += 1
        found.append(word)

print("\nSpam Keywords Found:", found)

if count >= 2:
    print("Result: Spam Email")
else:
    print("Result: Not Spam")
```
